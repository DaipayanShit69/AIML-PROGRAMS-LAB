```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(start):
    visited = []
    queue = [start]

    while queue:
        node = queue.pop(0)
        if node not in visited:
            visited.append(node)
            queue += graph[node]

    return visited

def dfs(node, visited=[]):
    visited.append(node)
    for x in graph[node]:
        if x not in visited:
            dfs(x, visited)
    return visited

start = input("Enter starting node: ").upper()

print("BFS:", bfs(start))
print("DFS:", dfs(start))
```
